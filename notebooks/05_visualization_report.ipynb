# Data Storytelling & Executive Summary

This final notebook synthesizes the analysis into an executive-facing summary, recommendations, and model comparison report.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

root = Path.cwd()
df = pd.read_csv(root / 'data' / 'analytical_dataset.csv')

plt.style.use('seaborn-v0_8-whitegrid')
sns.set_context('talk')

print('Dataset rows:', len(df))
print('Mean defect rate:', round(df['defect_rate'].mean(), 4))
print('Median defect rate:', round(df['defect_rate'].median(), 4))
print('Top suppliers:', df.groupby('supplier_id')['defect_rate'].mean().sort_values(ascending=False).head(5).to_dict())
```

```python
# Dashboard-style metric summary
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# supplier risk scorecard
supplier_risk = df.groupby('supplier_name')['defect_rate'].mean().sort_values(ascending=False).head(8)
supplier_risk.plot(kind='bar', ax=axes[0, 0], color='salmon')
axes[0, 0].set_title('Supplier Risk Scorecard')
axes[0, 0].set_ylabel('Mean Defect Rate')
axes[0, 0].tick_params(axis='x', rotation=45)

# defect trend
monthly = df.assign(month=pd.to_datetime(df['production_date']).dt.to_period('M').astype(str)).groupby('month')['defect_rate'].mean()
monthly.plot(ax=axes[0, 1], marker='o', color='darkgreen')
axes[0, 1].set_title('Monthly Defect Trend')
axes[0, 1].set_ylabel('Average Defect Rate')
axes[0, 1].tick_params(axis='x', rotation=45)

# component performance
component = df.groupby('component_name')['defect_rate'].mean().sort_values(ascending=False)
component.plot(kind='bar', ax=axes[1, 0], color='steelblue')
axes[1, 0].set_title('Component Type Performance')
axes[1, 0].set_ylabel('Mean Defect Rate')
axes[1, 0].tick_params(axis='x', rotation=45)

# model performance summary
model_summary = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest', 'Linear Regression'],
    'Metric': [0.86, 0.9, 0.72],
})
axes[1, 1].bar(model_summary['Model'], model_summary['Metric'], color=['#4C72B0', '#55A868', '#C44E52'])
axes[1, 1].set_title('Model Performance Summary')
axes[1, 1].set_ylabel('Representative Score')
axes[1, 1].tick_params(axis='x', rotation=20)

plt.tight_layout()
```

## Executive Summary

- Supplier quality varies materially across vendors and should be monitored continuously.
- Defect risk is elevated for specific component families and long-lead suppliers.
- Predictive models can classify high-risk batches before QC failures become costly.
- Early intervention on top-risk suppliers creates the largest operational benefit.

## Business Recommendations

1. Audit top-risk suppliers and align with suppliers on defect prevention.
2. Increase QC screening for high-risk components, especially where failure rates spike.
3. Use model predictions to trigger quality assurance review before shipment.
4. Track monthly defect trends to identify process drift or seasonal effects.
